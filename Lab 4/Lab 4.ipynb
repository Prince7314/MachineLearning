### How does `train_test_split` work? Show the code for splitting `iphone.csv` with a test size of 0.33.**
We separate our features ($X$, inputs) and targets ($y$, output/label) first, then split them using a test size of 0.33.

```python
X = df[['number']]
y = df['Price']
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.33,random_state=42)
```

## Step 5 . Convert state Column into Numeric Column

## Step 5.1 .  Perform Transformation

## Step 6 . Dummy variable trap

## Step 7 Splitting dataset in to Train and Test

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(x1,y,test_size=0.2,random_state=42)
```

## Step 8  Import LinearRegression model from linear_model family 

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

## Step 9  Fit the data

## Step 10  Predict the data

```python
y_predict = model.predict(x_test)
```
