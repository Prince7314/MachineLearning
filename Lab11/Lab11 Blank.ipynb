# Step 1: Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
df[['math score', 'reading score', 'writing score']].hist(
    figsize=(10, 5)
)

plt.tight_layout()
plt.show()
```

# Step 7: Apply StandardScaler

```python
X = df[['math score', 'reading score']]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

# Step 8: Elbow method to find optimal k

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
plt.figure(figsize=(8,5))
plt.plot(range(1,11), wcss, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("WCSS")
plt.title("Elbow Method")
plt.show()
```

# Step 9:Based on the elbow plot, choose an appropriate k value (e.g., k=3)

```python
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

clusters = kmeans.fit_predict(X_scaled)

df['Cluster'] = clusters

df[['math score',
    'reading score',
    'writing score',
    'Cluster']].head()
```

# Step 10: Print Cluster Center

```python
centers = scaler.inverse_transform(
    kmeans.cluster_centers_
)

print(centers)
```

# Step 11: Plot Cluster

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df['math score'],
    df['reading score'],
    c=df['Cluster'],
    cmap='viridis',
)

plt.xlabel("Math Score")

plt.ylabel("Reading Score")
plt.title("Student Clusters (K-Means)")
plt.show()
```

# Step 12:Analyze clusters 

```python
cluster_analysis = df.groupby('Cluster')[[
    'math score',
    'reading score',
]].mean()

print(cluster_analysis)
```
