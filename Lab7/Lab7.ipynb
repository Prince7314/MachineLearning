# Decision Tree Classifier

```python
import numpy as np
from sklearn.datasets import load_iris
```

# Import iris data ser using sklearn

```python
iris = load_iris()
display(iris)
```

# Importing the libraries

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
```

# Divide the data into input and output

```python
X = iris.data
y = iris.target
```

# Splitting the dataset into the Training set and Test set

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)
```

# Fitting DecisionTreeClassifier   on dataset

```python
dtree = DecisionTreeClassifier()
dtree.fit(X_train, y_train)
```

# Display Decision Tree

```python
plt.figure(figsize=(15, 20))
plot_tree(dtree, 
          filled=True, 
          feature_names=iris.feature_names,
          class_names=iris.target_names
)
plt.show()
```

# Predict the x_test

```python
y_predict = dtree.predict(X_test)
display(y_predict)
```

# Display Training Accuracy

```python
train_accuracy = dtree.score(X_train,y_train)
print(f"Training Accuracy: {train_accuracy}")
```

# Display Test Accuracy

```python
from sklearn.metrics import accuracy_score
acc = accuracy_score(y_test, y_predict)
print(f"Test Accuracy: {acc}")
```

# Confusion Matrix

```python
# print("Classification Report:")
# print(classification_report(y_test, y_predict))
```

# Save Model State

```python
import pickle
filename = 'decision_tree_model.pkl'
pickle.dump(dtree, open(filename, 'wb'))
print(f"Model saved to {filename}")
```

# Load Model from Disk

```python
loaded_model = pickle.load(open('decision_tree_model.pkl', 'rb'))
print("Model loaded successfully!")
```

```python
loaded_model_pred = loaded_model.predict(X_test)
print("Predictions from loaded model:")
display(loaded_model_pred)
```

```python
loaded_model_accuracy = accuracy_score(y_test, loaded_model_pred)
print(f"Accuracy of loaded model: {loaded_model_accuracy}")
```
